```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv("/content/student_performance_dataset (2).csv")
print("first 5 rows of dataset:")
print(df.head())
```

```python
print("Describing data :")
print(df.info())
```

```python
df.loc[0:5,'study_hours_per_week']= None
df.loc[10:15,'Attendence_rate']= None
df.loc[20:25,'past_exam_score']=None
df.loc[30:35,'final_exam_score']=None
print("\nmissing values before handling:")
print("df.isnull.sum()[df.isnull().sum()>0]")
```

```python
df['Study_Hours_per_Week']=df['Study_Hours_per_Week'].fillna(df['Study_Hours_per_Week'].mean())


df['Attendance_Rate']=df['Attendance_Rate'].fillna(df['Attendance_Rate'].median())


df['final_exam_score']=df['final_exam_score'].fillna(df['final_exam_score'])
```

```python
df=pd.get_dummies(df,columns=['Parental_Education_Level'])
print("\nOne-hot encoded coloumns:")
print(df.columns.tolist())
```

```python
le=LabelEncoder()
df['Pass_Fail_Encoded']=le.fit_transform(df['Pass_Fail'])

print("\nLabel Encoded Result:")
print(df[['Pass_Fail','Pass_Fail_Encoded']].head())
```

```python
df['Gender_Encoder'] = le.fit_transform(df['Gender'])
df['Internet_Access_Encoder'] = le.fit_transform(df['Internet_Access_at_Home'])
df['Extracurricular_Activities_Encoder'] = le.fit_transform(df['Extracurricular_Activities'])
```

```python
df.to_csv("student_performance_cleaned.csv",index=False)
print("\nCleaned dataset saved as 'student_performance_cleaned.csv'")
```
